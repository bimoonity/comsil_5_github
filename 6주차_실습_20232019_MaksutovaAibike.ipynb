# 코드 예시) Random 함수 


```python
import random
```

```python
#random.random() -> 0.0에서 1.0사이의 실수 중에서 난수값 리턴
print(random.random()) 
# ex) 0.2664520652284096

#random.uniform(a, b) -> 괄호 안 두 수 사이의 실수 중에서 난수값을 리턴
print(random.uniform(10, 30)) 
# ex) 28.038707445386397

#random.randint(a, b) -> 괄호 안 두 수 사이의 정수 중에서 난수값을 리턴
print(random.randint(100,200))
# ex) 106

#random.choice(sample) 함수 -> sample에서 무작위로 하나를 선택하여 리턴
data = [1, 2, 3, 4, 5, 6, 7]
print(random.choice(data))
# ex) 5

#random.sample(sample, n) -> 입력으로 받은 sample 에서 정한 개수만큼  무작위로 뽑아 리턴함
data = [1, 2, 3, 4, 'apple', 'banna']
print(random.sample(data, 3))
# ex) ['apple', 2, 3]
print(random.sample(data, 3))
# ex) ['apple', 3, 'banna']
print(random.sample(data, 5))
# ex) [2, 3, 1, 'apple', 4]
```

# 실습 2. Linear Congruential Generator(LCG) 구현  

X_(n+1) = (aX_n+C) mod m

LCG는 다음과 같은 인자들로 유일하게 결정됨  
  - 0 < m, 0 < a < m, 0 <= c < m, 0<=초기값 < m  

최대 주기 m을 가지기 위한 필요충분조건  

  - c와 m이 서로소, a-1이 m의 모든 소인수로 나뉨 
  - m이 4의 배수면 a-1도 4의 배수

```python
def lcg(seed, loop, m=2**32, a=214013, c=2531011):
    for i in range(loop):
        # write answer
        seed = (a * seed + c) % m   # X_(n+1) = (a * X_n + c) mod m
        yield seed
```

```python
num_iterations = 100
random_generator = lcg(seed=13, loop=num_iterations, m=2**32)
random_integers = []

## write answer
for num in random_generator:
    random_integers.append(num)

fig = plt.figure()
plt.hist(random_integers)
plt.title(f"Check Uniform Distribution of {num_iterations} iterations")
plt.xlabel("Numbers")
plt.ylabel("N of each number")
plt.xlim([0, 2**32])
plt.show()
```

# 실습 2-2 난수들 중 Uniform Distribution 확인  


```python
# LCG
from collections import defaultdict

m = 2**32
for num_iterations in [1000,10000,100000]:
    ## write answer
    counts = defaultdict(int)                       # 0~99 각 숫자가 나온 횟수
    random_generator = lcg(seed=13, loop=num_iterations, m=m)
    expected = num_iterations / 100                 # uniform이면 각 숫자가 나와야 하는 횟수
    for _ in range(num_iterations):
        ## write answer
        num = next(random_generator) * 100 // m     # [0, m) -> [0, 99]
        counts[num] += 1

    # 각 숫자의 (실제 횟수 - 기대 횟수) 차이의 평균 비율 (%)
    error = sum(abs(counts[k] - expected) for k in range(100)) / 100 / expected * 100
    print(f"LCG 시행횟수 {num_iterations}에서의 uniform distribution 오차 : {error}")
```

```python
# 메르센 트위스터 
from collections import defaultdict

for num_iterations in [1000,10000,100000]:
    ## write answer
    counts = defaultdict(int)
    expected = num_iterations / 100
    for _ in range(num_iterations):
        ## write answer
        num = random.randint(0, 99)                 # 파이썬 random = Mersenne Twister
        counts[num] += 1

    error = sum(abs(counts[k] - expected) for k in range(100)) / 100 / expected * 100
    print(f"메르센 트위스터 시행횟수 {num_iterations}에서의 uniform distribution 오차 : {error}")
```

# 실습 3. 주사위 합이 8인 경우

```python
# 출력시 컬러 지원 가능 모듈 설치
!pip install termcolor
from termcolor import colored
!pip install colorama
from colorama import init, Fore, Back, Style
```

```python
def roll_dice():
    ## write answer
    dice1 = random.randint(1, 6)
    dice2 = random.randint(1, 6)
    return dice1, dice2
```

```python
dice_tries = []
num_iterations = 100
hits = 0
for _ in range(num_iterations):
    ## write answer
    dice1, dice2 = roll_dice()
    dice_tries.append((dice1, dice2))
    if dice1 + dice2 == 8:
        hits += 1
print("="*100)
print(colored("* 두 주사위의 합이 8인 경우에는 cyan 색의 음영이 들어가 있음 *", 'red'))
for i in range(num_iterations):
  ## write answer
  dice1, dice2 = dice_tries[i]
  text = f"{i+1}번째 : {dice1} + {dice2} = {dice1 + dice2}"
  if dice1 + dice2 == 8:
    print(colored(text, on_color='on_cyan'))
  else:
    print(text)


print(colored("\n실제 값 : 0.138889", "red"))
print(colored(f"계산된 값 : {round(hits / num_iterations,6)}", "red"))
print(colored(f"오차율 : {abs(hits / num_iterations - 5/36) / (5/36) * 100} %", "red"))
print("="*100)
```

# 코딩 과제 6. 원주율 구하기

```python
import math

MAX_POINTS = 1000000
DRAW_POINTS = 5000      # 그림에는 앞의 5000개만 그림 (100만 개를 다 그리면 너무 느림)

inside = 0
in_x, in_y, out_x, out_y = [], [], [], []

for i in range(MAX_POINTS):
    x = random.uniform(-1, 1)
    y = random.uniform(-1, 1)
    if x*x + y*y <= 1:              # 반지름 1인 원 안
        inside += 1
        if i < DRAW_POINTS:
            in_x.append(x); in_y.append(y)
    else:
        if i < DRAW_POINTS:
            out_x.append(x); out_y.append(y)

# (원의 넓이) / (정사각형 넓이) = pi / 4
pi = 4 * inside / MAX_POINTS
error = abs(pi - math.pi) / math.pi * 100
print(f"계산된 pi : {pi}")
print(f"실제 pi   : {math.pi}")
print(f"오차율    : {error} %")

fig = plt.figure(figsize=(6, 6))
plt.scatter(in_x, in_y, s=1, c='red', label='inside')
plt.scatter(out_x, out_y, s=1, c='blue', label='outside')
plt.gca().set_aspect('equal')
plt.title(f"Monte Carlo pi = {pi}")
plt.legend(loc='upper right')
plt.show()
```
